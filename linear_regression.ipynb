```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([[1], [2], [3], [4]])
y = np.array([2, 4, 6, 8])

model = LinearRegression()
model.fit(X, y)

print("Slope:", model.coef_)
print("Intercept:", model.intercept_)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression

# Step 1: Create dataset
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)   # Independent variable
y = np.array([35, 40, 50, 60, 65])            # Dependent variable

# Step 2: Create model
model = LinearRegression()

# Step 3: Train model
model.fit(X, y)

# Step 4: Predict for 6 hours
prediction = model.predict([[6]])

# Step 5: Print results
print("Predicted Marks for 6 hours:", prediction[0])
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)   # Experience
y = np.array([3, 4, 5, 6, 7])                  # Salary
model = LinearRegression()
model.fit(X, y)
prediction = model.predict([[6]])
print("Predicted Salary for 6 years experience:", prediction[0])
print("Slope (Coefficient):", model.coef_[0])
print("Intercept:", model.intercept_)
```
