```python
import numpy as np
from sklearn.tree import DecisionTreeClassifier

# Data
X = np.array([1,2,3,4,5,6]).reshape(-1,1)
y = np.array([0,0,0,1,1,1])

# Model
dt = DecisionTreeClassifier()
dt.fit(X, y)

# Prediction
print("Prediction for 3.5 hours:", dt.predict([[3.5]]))
```

```python
new_data = np.array([2,4,6]).reshape(-1,1)
print("Predictions:", dt.predict(new_data))
```

```python
dt_entropy = DecisionTreeClassifier(criterion="entropy")
dt_entropy.fit(X, y)

print("Prediction:", dt_entropy.predict([[3.5]]))
```

```python
dt_limited = DecisionTreeClassifier(max_depth=2)
dt_limited.fit(X, y)

print("Prediction:", dt_limited.predict([[5]]))
```

```python
from sklearn.tree import DecisionTreeRegressor

X = np.array([1,2,3,6,7]).reshape(-1,1)
y = np.array([10,12,14,20,22])

dt_reg = DecisionTreeRegressor()
dt_reg.fit(X, y)

print("Predicted value for X=5:", dt_reg.predict([[5]]))
```
